### Create Class For Circle and Rectangle

```python
class Circle:
    def __init__(self,r):
        self.r = r
       
    def Area(self):
        return 3.14*self.r*self.r
    
    def Perimeter(self):
        return 2*3.14*self.r
```

```python
c1 = Circle(3)
```

```python
c1.Area()
```

```python
c1.Perimeter()
```

```python
class Rectangle:
    def __init__(self,height,width):
        self.height = height
        self.width = width

    def perimeter(self):
        return 2*self.height + 2*self.width
    
    def Area():
        return self.height * self.width

       
```

```python
@dataclass
class Triangle :
    a : int|float
    b : int|float
    c : int|float
        
    def perimeter(self):
        if (
            (self.a + self.b > self.c) and 
            (self.b + self.c > self.a)  and 
            (self.c + self.a > self.b) ):
             return self.a + self.b + self.c
             
      
    def Area(self):
        if self.perimeter() is not None:
            s = self.perimeter()/2
            return (s*(s-self.a)*(s-self.b)*(s-self.c))**(1/2)
        else:
            return 'Perimeter is None'
```

```python
t = Triangle(23,34,45)
t.perimeter()
```
