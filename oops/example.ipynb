```python
from dataclasses import dataclass
```

```python
@dataclass
class Triangle :
    a : int|float
    b : int|float
    c : int|float

    def is_triangle(self):
        if (
            (self.a + self.b > self.c) and 
            (self.b + self.c > self.a)  and 
            (self.c + self.a > self.b) ):
            return True
        else:
            return False
        
    def perimeter(self):
        return self.a + self.b + self.c
    def Area(self):
        s = self.perimeter()/2
        return (s*(s-self.a)*(s-self.b)*(s-self.c))**(1/2)
```

```python
# create class for simple calculator
- Addition
-Substraction
-Multiplication
-Division
```

```python
@dataclass
class Calculator:
    x:int|float
    y:int|float

    def addition(self):
        return self.x + self.y
    
    def substraction(self):
        return self.x - self.y
    
    def multiplication(self):
        return self.x * self.y
    
    def division(self):
        if self.y!= 0:
            return self.x/self.y
        else:
            raise ValueError (Infinity)

```
